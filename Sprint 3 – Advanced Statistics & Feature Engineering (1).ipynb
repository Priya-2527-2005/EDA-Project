#  2. Hypothesis Testing


### Perform appropriate hypothesis tests wherever applicable.
#### Examples:


## ➤Z-Test


### Define the hypotheses

#### 1- Null Hypothesis (H₀):The average arrival delay is 0 minutes.
#### 2- Alternate Hypothesis (H₁):The average arrival delay is different from 0 minutes.
#### 3- Significance Level (α):0.05

### Perform the Z-Test

```python
from statsmodels.stats.weightstats import ztest

arrival_delay = df['ARRIVAL_DELAY'].dropna()

z_stat, p_value = ztest(
    arrival_delay,
    value=0,
    alternative='two-sided'
)

print("Z-Statistic:", round(z_stat, 4))
print("P-Value:", round(p_value, 6))
```

### Decision

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

#### 1- Since the p-value (0.0) is less than the significance level (0.05), we reject the Null Hypothesis.
#### 2- There is statistically significant evidence that the average arrival delay is different from 0 minutes.

### Business Interpretation

#### 1- The average arrival delay is statistically significant.
#### 2- Flights are not consistently arriving exactly on schedule.
#### 3- Airlines should focus on reducing operational delays.
#### 4- Improving departure and turnaround processes can help improve arrival punctuality.

## ➤T-Test

### Hypotheses

#### 1- Null Hypothesis (H₀):The average arrival delay is 0 minutes.
#### 2- Alternate Hypothesis (H₁):The average arrival delay is different from 0 minutes.
#### 3- Significance Level (α):0.05

### Perform the T-Test

```python
from scipy.stats import ttest_1samp

arrival_delay = df['ARRIVAL_DELAY'].dropna()

t_stat, p_value = ttest_1samp(
    arrival_delay,
    popmean=0
)

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 6))
```

### Decision

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

#### 1- The p-value (0.0) is less than 0.05.
#### 2- Therefore, we reject the Null Hypothesis.
#### 3- There is statistically significant evidence that the average arrival delay differs from 0 minutes.

### Business Interpretation

#### 1- Average arrival delay is statistically significant.
#### 2- Flight punctuality differs from the on-time target.
#### 3- Airlines should focus on reducing operational delays.
#### 4- Improving departure and turnaround processes can improve arrival performance.

## ➤Chi-Square Test

### Create the contingency table

```python
print(df.columns.tolist())
```

```python
def delay_category(delay):
    if delay <= 0:
        return 'On Time/Early'
    elif delay <= 15:
        return 'Minor Delay'
    elif delay <= 60:
        return 'Moderate Delay'
    else:
        return 'Severe Delay'

df['DELAY_CATEGORY'] = df['ARRIVAL_DELAY'].apply(delay_category)
```

```python
df['DELAY_CATEGORY'].value_counts()
```

```python
from scipy.stats import chi2_contingency

crosstab = pd.crosstab(
    df['AIRLINE'],
    df['DELAY_CATEGORY']
)

crosstab
```

```python
chi2, p_value, dof, expected = chi2_contingency(crosstab)

print("Chi-Square Statistic:", round(chi2, 4))
print("P-Value:", round(p_value, 6))
print("Degrees of Freedom:", dof)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

#### 1- The p-value (0.0) is less than 0.05.
#### 2- Therefore, we reject the Null Hypothesis.
#### 3- There is statistically significant evidence of an association between airline and delay category.

### Business Interpretation

#### 1- Delay patterns differ significantly across airlines.
#### 2- Some airlines may experience higher proportions of delayed flights.
#### 3- Airlines should identify the operational factors behind these differences.
#### 4- Airline-specific delay management can improve overall flight performance.

## ➤ANOVA

### Hypotheses

#### 1- Null Hypothesis (H₀):All airlines have the same average arrival delay.
#### 2- Alternate Hypothesis (H₁):At least one airline has a different average arrival delay.
#### 3- Significance Level (α):0.05

### Perform One-Way ANOVA

```python
from scipy.stats import f_oneway

airline_groups = [
    group['ARRIVAL_DELAY'].dropna()
    for _, group in df.groupby('AIRLINE')
]

f_stat, p_value = f_oneway(*airline_groups)

print("F-Statistic:", round(f_stat, 4))
print("P-Value:", round(p_value, 6))
```

### Decision

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

#### 1- The p-value (0.0) is less than 0.05.
#### 2- Therefore, we reject the Null Hypothesis.
#### 3- There is statistically significant evidence that average arrival delays differ among airlines.

### Business Interpretation

#### 1- Arrival delay performance differs significantly across airlines.
#### 2- Some airlines have higher average delays than others.
#### 3- Airlines should identify the operational causes of these differences.
#### 4- Airline-specific delay reduction strategies can improve punctuality.

# State:

## ➤Null Hypothesis

#### 1- The Null Hypothesis assumes there is no significant difference between the groups.
#### 2- In our ANOVA, it assumes all airlines have the same average arrival delay.
#### 3- Any observed difference is considered due to random variation.
#### 4- We test this assumption using the ANOVA test.
#### 5- If the p-value is less than 0.05, we reject the Null Hypothesis.

## ➤Alternate Hypothesis

#### 1- The Alternate Hypothesis assumes there is a significant difference between the groups.
#### 2- In our ANOVA, at least one airline has a different average arrival delay.
#### 3- This means airline performance may affect arrival delays.
#### 4- The observed differences are unlikely to be due only to random variation.
#### 5- If the p-value is less than 0.05, we reject the Null Hypothesis in favor of the Alternate Hypothesis.

## ➤Test Statistic


#### 1- The test statistic is the value calculated from the statistical test.
#### 2- It measures how strongly the sample results differ from the Null Hypothesis.
#### 3- In our One-Way ANOVA, the test statistic is called the **F-Statistic**.
#### 4- Our calculated F-Statistic is **20.3707**.
#### 5- A larger F-Statistic indicates greater differences between airline average arrival delays.

## ➤p-value

#### 1- The p-value shows how likely the observed result is if the Null Hypothesis is true.
#### 2- A smaller p-value indicates stronger evidence against the Null Hypothesis.
#### 3- In our ANOVA, the p-value is **0.0**.
#### 4- Since **0.0 < 0.05**, the result is statistically significant.
#### 5- Therefore, we reject the Null Hypothesis.

## ➤Decision

#### 1- The significance level is **0.05**.
#### 2- The p-value is **0.0**.
#### 3- Since **p-value < 0.05**, we reject the Null Hypothesis.
#### 4- There is a significant difference in average arrival delays among airlines.
#### 5- Therefore, airline performance is significantly associated with arrival delay.

## ➤Business Interpretation

#### 1- Average arrival delays differ significantly among airlines.
#### 2- Some airlines may have better on-time performance than others.
#### 3- Airlines should identify the operational causes of higher delays.
#### 4- Airline-specific delay management can improve punctuality.

# 3. Confidence Interval Analysis
### Construct confidence intervals for appropriate numerical features.
### Interpret the results.


```python
from scipy.stats import t

arrival_delay = df['ARRIVAL_DELAY'].dropna()

mean = arrival_delay.mean()
std_error = arrival_delay.sem()
confidence = 0.95

ci = t.interval(
    confidence,
    df=len(arrival_delay) - 1,
    loc=mean,
    scale=std_error
)

print("Mean Arrival Delay:", round(mean, 2))
print("95% Confidence Interval:", 
      round(ci[0], 2), "to", round(ci[1], 2))
```

#### 1- The mean arrival delay is **4.00 minutes**.
#### 2- The 95% confidence interval ranges from **3.67 to 4.33 minutes**.
#### 3- We are 95% confident that the true average arrival delay lies within this range.
#### 4- The interval is relatively narrow, indicating a precise estimate.
#### 5- This suggests that the average arrival delay is consistently around 4 minutes.

#### 2- Peak-hour flights can create additional pressure on airport resources.
#### 3- Busy airports may require better coordination and scheduling.
#### 4- Flight distance and operational conditions should be considered when planning schedules.
#### 5- Delay-risk information can help airlines identify flights that may require additional planning.

### – Practical Recommendations

#### 1- Avoid scheduling too many flights during peak operating hours.
#### 2- Add sufficient buffer time between tightly scheduled flights.
#### 3- Allocate more aircraft, crew, and ground staff during busy periods.
#### 4- Adjust schedules at airports with consistently high traffic and delays.
#### 5- Use historical delay patterns when planning flight timings.
#### 6- Give additional turnaround time to flights with higher delay risk.

## ➤Improve passenger satisfaction


#### 1- Flight delays can negatively affect the passenger experience.
#### 2- Severe and high-risk delays require greater attention.
#### 3- Better on-time performance can improve customer satisfaction.
#### 4- Clear communication during delays can reduce passenger frustration.
#### 5- Reliable scheduling helps passengers plan their travel with confidence.

### – Practical Recommendations

#### 1- Provide passengers with timely updates about delays and schedule changes.
#### 2- Send alerts through SMS, email, or mobile apps when delays occur.
#### 3- Offer quick rebooking options for significantly delayed or cancelled flights.
#### 4- Provide appropriate assistance to passengers during long delays.
#### 5- Improve punctuality to reduce waiting time and travel disruptions.
#### 6- Use passenger feedback to identify and improve common service issues.

## ➤Reduce cancellation rates

#### 1- Cancellation reasons should be monitored to identify the major causes of cancellations.
#### 2- Weather, airline, and operational issues can affect flight reliability.
#### 3- High cancellation periods should be identified through historical data.
#### 4- Better scheduling and resource planning can help reduce avoidable cancellations.
#### 5- Reducing cancellations improves operational reliability and passenger satisfaction.

### – Practical Recommendations

#### 1- Analyze cancellation reasons regularly and address the most frequent causes.
#### 2- Keep backup aircraft and crew available during high-risk periods.
#### 3- Improve weather monitoring and prepare alternative operating plans.
#### 4- Avoid overloading busy airports with tightly scheduled flights.
#### 5- Provide early rebooking options when cancellation becomes likely.
#### 6- Maintain regular aircraft inspections to reduce maintenance-related cancellations.
