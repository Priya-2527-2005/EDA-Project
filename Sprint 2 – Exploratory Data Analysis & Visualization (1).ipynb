## ➤Outliers  

```python
Q1 = df[numerical_cols].quantile(0.25)
Q3 = df[numerical_cols].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_count = ((df[numerical_cols] < lower_bound) |
                 (df[numerical_cols] > upper_bound)).sum()

print("Outlier count:")
print(outlier_count)
```

```python
import matplotlib.pyplot as plt

df[numerical_cols].boxplot(figsize=(20, 8))
plt.xticks(rotation=90)
plt.title("Box Plot of Numerical Variables")
plt.ylabel("Values")
plt.tight_layout()
plt.show()
```

#### - Outliers were identified using the **Interquartile Range (IQR) method**.
#### - Box plots were used to visually inspect the spread and extreme values.
#### - Delay-related variables show noticeable extreme values due to flights experiencing unusually high delays.
#### - The identified outliers are retained because they represent genuine flight operational events and may provide useful business insights.

## ➤Airline vs Arrival Delay


### Calculate average arrival delay

```python
airline_arrival_delay = (
    df.groupby('AIRLINE')['ARRIVAL_DELAY']
      .mean()
      .sort_values(ascending=False)
)

print(airline_arrival_delay)
```

#### - Average arrival delay was calculated for each airline.
#### - The analysis shows differences in arrival delay performance among airlines.
#### - Airlines with higher average arrival delays may require closer attention to operational efficiency.

## 5. Group-Based Analysis

## Perform grouped analysis using appropriate categorical features.
### Generate meaningful summaries by comparing:


## ➤Airlines

```python
airline_summary = df.groupby('AIRLINE').agg(
    Total_Flights=('AIRLINE', 'count'),
    Avg_Departure_Delay=('DEPARTURE_DELAY', 'mean'),
    Avg_Arrival_Delay=('ARRIVAL_DELAY', 'mean'),
    Cancellation_Rate=('CANCELLED', 'mean')
)

airline_summary['Cancellation_Rate'] *= 100

print(airline_summary)
```

#### - Flight performance was compared across different airlines.
#### - Average departure and arrival delays were calculated for each airline.
#### - Cancellation rates were also compared to identify operational differences.
#### - This helps identify airlines with better or poorer operational performance.

## ➤Airports

```python
airport_summary = df.groupby('ORIGIN_AIRPORT').agg(
    Total_Flights=('ORIGIN_AIRPORT', 'count'),
    Avg_Arrival_Delay=('ARRIVAL_DELAY', 'mean')
).sort_values('Avg_Arrival_Delay', ascending=False)

print(airport_summary.head(10))
```

#### - Airports were compared based on the number of flights and average arrival delay.
#### - The airports with higher average delays were identified.
#### - This helps highlight airports that may experience operational congestion.
